```python
import pandas as pd

df = pd.read_excel("../data/raw/Online_Retail.xlsx")

print("Before:", df.shape)

df = df.dropna(subset=["CustomerID"])

df = df[~df["InvoiceNo"].astype(str).str.startswith("C")]

df = df[(df["Quantity"] > 0) & (df["UnitPrice"] > 0)]

df = df.drop_duplicates()

print("After:", df.shape)

df.head()
```

```python
basket = (
    df.groupby(['InvoiceNo', 'Description'])['Quantity']
    .sum()
    .unstack()
    .fillna(0)
)

basket.head()
```

```python
basket = basket.applymap(lambda x: 1 if x > 0 else 0)

basket.head()
```

```python
from mlxtend.frequent_patterns import apriori, association_rules
```

```python
frequent_items = apriori(
    basket,
    min_support=0.02,
    use_colnames=True
)

frequent_items.head()
```

```python
rules = association_rules(
    frequent_items,
    metric="confidence",
    min_threshold=0.3
)

rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']].head()
```

```python
top_rules = rules.sort_values(by='lift', ascending=False)

top_rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']].head(10)
```

```python
print("Number of generated rules:", len(rules))
```

```python
rules[['support', 'confidence', 'lift']].describe()
```

```python
from mlxtend.frequent_patterns import fpgrowth

fp_items = fpgrowth(
    basket,
    min_support=0.02,
    use_colnames=True
)

fp_rules = association_rules(
    fp_items,
    metric="confidence",
    min_threshold=0.3
)

print("FP-Growth itemsets:", len(fp_items))
print("FP-Growth rules:", len(fp_rules))

fp_rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']].sort_values(
    by="lift", ascending=False
).head(10)
```

```python
print("FP-Growth itemsets:", len(fp_items))
print("FP-Growth rules:", len(fp_rules))
```

```python
from itertools import combinations

vertical_data = {}

for product in basket.columns:
    vertical_data[product] = set(basket.index[basket[product] == 1])

num_transactions = len(basket)

min_support = 0.02
min_count = int(min_support * num_transactions)

frequent_1 = {
    frozenset([product]): tids
    for product, tids in vertical_data.items()
    if len(tids) >= min_count
}

print("Frequent 1-itemsets:", len(frequent_1))
```

```python
eclat_itemsets = []

items = list(frequent_1.items())

for i in range(len(items)):
    itemset1, tids1 = items[i]

    for j in range(i + 1, len(items)):
        itemset2, tids2 = items[j]

        combined_itemset = itemset1 | itemset2
        combined_tids = tids1 & tids2

        support = len(combined_tids) / num_transactions

        if support >= min_support:
            eclat_itemsets.append({
                "support": support,
                "itemsets": combined_itemset
            })

eclat_df = pd.DataFrame(eclat_itemsets)

print("ECLAT itemsets:", len(eclat_df))

eclat_df.sort_values(by='support', ascending=False).head(10)
```

```python
eclat_rules = []

support_dict = {}

for itemset, tids in frequent_1.items():
    support_dict[itemset] = len(tids) / num_transactions

for _, row in eclat_df.iterrows():
    support_dict[row["itemsets"]] = row["support"]

for _, row in eclat_df.iterrows():
    itemset = row["itemsets"]
    itemset_support = row["support"]

    for item in itemset:
        antecedent = frozenset([item])
        consequent = itemset - antecedent

        antecedent_support = support_dict.get(antecedent, 0)
        consequent_support = support_dict.get(consequent, 0)

        if antecedent_support > 0 and consequent_support > 0:
            confidence = itemset_support / antecedent_support
            lift = confidence / consequent_support

            if confidence >= 0.3:
                eclat_rules.append({
                    "antecedents": antecedent,
                    "consequents": consequent,
                    "support": itemset_support,
                    "confidence": confidence,
                    "lift": lift
                })

eclat_rules_df = pd.DataFrame(eclat_rules)

print("ECLAT rules:", len(eclat_rules_df))

eclat_rules_df.sort_values(by="lift", ascending=False).head(10)
```

```python
comparison = pd.DataFrame({
    "Algorithm": ["Apriori", "FP-Growth", "ECLAT"],

    "Frequent Itemsets": [
        len(frequent_items),
        len(fp_items),
        len(eclat_df)
    ],

    "Generated Rules": [
        len(rules),
        len(fp_rules),
        len(eclat_rules_df)
    ],

    "Max Confidence": [
        rules["confidence"].max(),
        fp_rules["confidence"].max(),
        eclat_rules_df["confidence"].max()
    ],

    "Max Lift": [
        rules["lift"].max(),
        fp_rules["lift"].max(),
        eclat_rules_df["lift"].max()
    ]
})

comparison
```
